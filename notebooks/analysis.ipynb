```python
import numpy as np
import pandas as pd
from pathlib import Path
```

```python
cwd=Path.cwd()
parrent_dir=cwd.parent
print(parrent_dir)
# Read the csv file
df = pd.read_csv(parrent_dir/'data-scraped'/'appartment_cleaned_interim.csv')
df.head(10)
```

```python
df.isnull().sum()
```

```python
df.dropna(subset=['nearbylocation','facility'], inplace=True)
```

```python
df.isnull().sum()
```

```python
df.shape
```

#carpet to built up ratio:  0.8416612442582418
#super to built up ratio:  1.1127433971009146


```python
# both super is null and built up present built up null

sc_df = df[df['super_bulit_area'].isnull()]
```

```python
sc_df.loc[:,'super_bulit_area'] = round(sc_df['bulit_area'] * 1.127,2)
```

```python
df.update(sc_df)
```

```python
df.isnull().sum()
```

```python
loc_df = pd.read_csv('loc_find.csv')
```

```python
loc_df.columns
```

```python
merged_df = pd.merge(
    df,
    loc_df,
    on=['apartment_name', 'appartment_loc'],  # Columns to join on
    how='inner'  # Inner join
)
```

```python
merged_df.head()
```

```python
merged_df.shape
```

```python
merged_df.drop(columns=['apartment_name', 'appartment_loc'], inplace=True)
```

```python
merged_df.rename(columns={
    'corrected_appartment_loc': 'apartment_loc',
    'corrected_apartment_name': 'apartment_name',
}, inplace=True)
```

```python
merged_df.isnull().sum()
```

```python
merged_df
```

```python
merged_df['price_per_sqft'] = (merged_df['price_value'] * 1e7) / merged_df['super_bulit_area']
```

```python
# Select only numeric columns for aggregation
numeric_columns = ['price_per_sqft', 'bulit_area', 'Latitude', 'Longitude']
grouped_df = merged_df.groupby('apartment_loc')[numeric_columns].mean()
grouped_df.head()
```
